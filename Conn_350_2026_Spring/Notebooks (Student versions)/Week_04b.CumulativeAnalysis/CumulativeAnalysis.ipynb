## Introduction
Here, we're going to focus on the accumulated sum of emissions over time, written mathematically as

$$
E(t) = \int_{t_0}^t F_{ha}(t) \ dt \ \ \ \ \ (1)
$$

where $F_{ha}(t)$ is the annual emission rate we created in ScheduledFlows, $t_0$ is a starting time (we'll make that the year 1750). Why would we care about this accumulation? There's a school of thought in climate science, that what really counts is not so much how much carbon goes into the air any given year, but total amount humans put there.

There's an easy way to do this in Python: we just add up the emissions year after year, making sure to correct for the fact that the time step in our emission schedule is some fraction of a year. Details are given below.
