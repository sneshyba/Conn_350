### Overview
Here we'll put together a climate model based on carbon mass balance involving the atmosphere, land, and the oceans. The model is based on the fluxes ($GtC/yr$) between these reservoirs,

$$
F_{la} =  k_{la} \ \ \ (1) 
$$

$$
F_{al} = k_{al0} +  k_{al1} \times C_{atm} \ \ \ (2)
$$

$$
F_{oa} = k_{oa} C_{ocean} \ \ \ (3)
$$

$$
F_{ao} = k_{ao} C_{atm} \ \ \ (4)
$$

$$
F_{ha} = \epsilon(t) \ \ \ (5)
$$

As previously, we'll solve this problem numerically by setting up a loop that adjusts the amount of carbon in the reservoirs at each time step, 

$$
C_{atm,n+1} = C_{atm,n}+(F_{la}+F_{oa}+F_{ha}-F_{al}-F_{ao}) \times \Delta t \ \ \ (6)
$$

$$
C_{ocean,n+1} = C_{ocean,n}+(F_{oa}-F_{ao}) \times \Delta t \ \ \ (7)
$$

where $C_{atm,n}$ means "amount of carbon in the atmosphere in the current step" and $C_{atm,n+1}$ means "amount of carbon in the atmosphere in the next step," and so on. In Python, we'll be implementing Eqs. 6 and 7 using the "+=" construct, which we encountered before in CumulativeAnalysis.

### Analysis
A key focus of the analysis we'll be interested in here is a metric called the *net flux* between reservoirs of carbon. The two we'll be most interested in are the land/atmosphere net flux, defined by

$$
F_{la,net}=F_{la} - F_{al} \ \ \ (8)
$$

and the ocean/atmosphere net flux,

$$
F_{oa,net}=F_{oa} - F_{ao} \ \ \ (9)
$$

The sign of these quantities tells us something really important: if negative, it means that the oceans and land are acting as *net sinks* (sometimes shortened to just *sinks*) for atmospheric carbon. Another metric we'll focus on is the strength of Earth's natural ecosystem service. These are the net fluxes just defined expressed as percentages of the anthropogenic emission in any given year:

$$
\% F_{la,eco}=F_{la,net}/F_{ha} \times 100 \ \ \ (10)
$$

$$
\% F_{oa,eco}=F_{oa,net}/F_{ha} \times 100 \ \ \ (11)
$$
